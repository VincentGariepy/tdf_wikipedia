```python
from pcs_parsing.pcs_stage_parsing import fetch_full_stage_data
from wikidata_functions.wikidata_upload_stage import update_stage_properties, upload_entity
```

```python
TDF_YEAR = 1947
```

```python
for stage_number in range(1, 6):
    results = fetch_full_stage_data(f"https://www.procyclingstats.com/race/tour-de-france/{str(TDF_YEAR)}/stage-{str(stage_number)}", stage_number, TDF_YEAR)
    stage_wikidata_id = results['stage_general_info']['stage_wikidata_id']

    entity = update_stage_properties(stage_wikidata_id, results)
    upload_entity(entity, stage_wikidata_id)
```

```python
results = fetch_full_stage_data(f"https://www.procyclingstats.com/race/tour-de-france/{str(TDF_YEAR)}/stage-{str(6)}", 6, TDF_YEAR)
stage_wikidata_id = results['stage_general_info']['stage_wikidata_id']

entity = update_stage_properties(stage_wikidata_id, results)
upload_entity(entity, stage_wikidata_id)
```
